# Accumulated variables, forecasting and data assimilation

To understand the challenges of forecasting accumulated variables like precipitation, we need to consider how forecasting models and data assimilation techniques work together. 

Specifically, we need to consider the following points:

- Some variables are accumulated over time, such as precipitation, i.e. the total amount produced in a time-period is computed. For this a forecasting model needs to produce a forecast over a time period so the variable can be accumulated. 
- At the same time we want to assimilate observations into the atmospheric state produced by the forecasting model, to produce a historical record.
- in ERA5 they use 4DVar for data assimilation and can therefore continually assimilate for every forecasting time step, while for DANRA 3DVar was used which only assimilates at initial time steps. And so there is a risk in DANRA that the forecast may drift away from reality over time
- the existance of clouds and precipitation are not assimilated, meaning that there are no clouds present in the initial condition of the forecast. Conventional wisdom (need ref) says that the forecasting model should therefore be run for a spin-up period to allow clouds and precipitation to develop before the forecast is considered reliable.

So in summary, precipitation is accumulated and therefore must be forecasted
over time, in DANRA forecasts may drift from the true atmospheric state because
observations are only assimilated into the beginning of the forecast, and in
the start of the forecast there are no clouds present and they must form before precipitation can occur.

This leaves us with the question of *over what forecasting horizon should we trust forecasts of accumulated precipititation?* Too early and the forecast may not have had enough time to develop clouds and precipitation, too late and the forecast may have drifted away from reality due to the lack of continuous assimilation.

As well as running forecasts for producing accumulated variables, we also ran forecasts for the purpose of producing a historical record with higher temporal resolution than we could afford to do assimilation at.

During the creation of DANRA we produced two sets of forecasts:

1. 18-hourly forecasts, started at 00 UTC and 12 UTC, with the expectation that after 18 hours the forecasts may have drifted away from reality due to the lack of continuous assimilation.
2. 3-hourly forecasts, started for every assimilation step, so every 3 hours (00 UTC, 03 UTC, 06 UTC, etc.). It was expected (from prior results) that these forecasts would not be reliable for precipitation, as they would not have had enough time to develop clouds and precipitation, but may have been more reliable for other variables like temperature and wind.

On comparing the two sets of forecasts with radar observations of precitipitation, we found that the **3-hourly forecasts were more reliable for precipitation than the 18-hourly forecasts**. This was surprising, as we had expected the 18-hourly forecasts to be more accurate based on prior experience, however given that convective clouds form on the order of tens of minutes to a few hours, it seems that the 3-hourly forecasts were able to capture the development of clouds and precipitation better than the 18-hourly forecasts, which may have drifted away from reality due to the lack of continuous assimilation.

It is however not given that the 3-hourly forecasts will always be more reliable than the 18-hourly forecasts for all variables. For example, for variables that are not accumulated, such as temperature or wind, the 18-hourly forecasts may be more reliable as they have had more time to develop and may not be as affected by the lack of continuous assimilation.

For this reason we have decided to share both the 3-hourly and 18-hourly forecasts in DANRA, so that users can choose which set of forecasts to use based on their specific needs and the variables they are interested in.
